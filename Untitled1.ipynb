```python
def teks_ke_mesin(teks):
    print(f"\nTeks asli: {teks}\n")
    print(f"{'Karakter':<10}{'ASCII':<8}{'Biner':<12}{'Heksa'}")
    print("-" * 40)

    hasil_biner = []
    for huruf in teks:
        ascii_val = ord(huruf)              # karakter -> angka ASCII
        biner = format(ascii_val, "08b")    # angka -> biner 8 bit
        heksa = format(ascii_val, "02X")    # angka -> heksadesimal
        hasil_biner.append(biner)
        label = huruf if huruf != " " else "(spasi)"
        print(f"{label:<10}{ascii_val:<8}{biner:<12}{heksa}")

    print("\nBahasa mesin (biner):")
    print(" ".join(hasil_biner))
    return hasil_biner


def mesin_ke_teks(biner_str):
    # kebalikannya: biner -> teks
    return "".join(chr(int(b, 2)) for b in biner_str.split())


# --- Program utama ---
print("=== KONVERTER BAHASA MANUSIA KE BAHASA MESIN ===")
teks = input("Masukkan kalimat: ")
biner = teks_ke_mesin(teks)

print("\nDikembalikan ke teks:", mesin_ke_teks(" ".join(biner)))
```
